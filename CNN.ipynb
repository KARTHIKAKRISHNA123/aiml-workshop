```python


# Create blank emoji (8x8 with RGB)
emoji = np.zeros((8, 8, 3), dtype=np.uint8)

# Apply gradient (yellow → orange)
for i in range(8):
    for j in range(8):
        emoji[i, j] = [255, 255 - (i * 10), 0]

# Eyes
emoji[2, 2] = [0, 0, 0]
emoji[2, 5] = [0, 0, 0]

# Curved smile
emoji[5, 2] = [0, 0, 0]
emoji[6, 3] = [0, 0, 0]
emoji[6, 4] = [0, 0, 0]
emoji[5, 5] = [0, 0, 0]

# Optional: blush
emoji[4, 1] = [255, 100, 100]
emoji[4, 6] = [255, 100, 100]

# Scale up for better visibility
emoji_big = np.kron(emoji, np.ones((40, 40, 1)))

# Display
plt.imshow(emoji_big.astype(np.uint8))
plt.axis('off')
plt.show()
```

```python
#Calculate the size of the image
image_size = emoji.size * emoji.itemsize * 3
print("Image Size in Bits!", image_size)

print("Image size in KB!", image_size / 1024)

```
